## 7. Data Cleaning and Preprocessing

The dataset was checked for missing values and incomplete observations.

Missing values in Rainfall, Soil Moisture, and Yield were handled using the
median value of the respective columns.

After preprocessing, the dataset contained no missing values.

```python
df.isnull().sum()
```

```python
df["Rainfall_mm"] = df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Yield_Tonnes_Ha"].median()
)
```

## 9. Statistical Analysis and Visualization

Statistical and visualization techniques were used to examine relationships,
differences, and variations in agricultural performance.

The analysis included correlation analysis, the Kruskal-Wallis test,
seasonal comparisons, state-season comparisons, heatmaps, bar charts, and
outlier analysis.

```python
from scipy.stats import kruskal

kharif = df[df["Season"] == "Kharif"]["Yield_Tonnes_Ha"]
rabi = df[df["Season"] == "Rabi"]["Yield_Tonnes_Ha"]
zaid = df[df["Season"] == "Zaid"]["Yield_Tonnes_Ha"]

stat, p_value = kruskal(kharif, rabi, zaid)

print("Kruskal-Wallis Statistic:", round(stat, 2))
print("p-value:", p_value)
```

```python
df.groupby("State")["Profit_INR"].mean().sort_values(ascending=False).round(2)
```

```python
state_season_yield = pd.pivot_table(
    df,
    index="State",
    columns="Season",
    values="Yield_Tonnes_Ha",
    aggfunc="mean"
)

state_season_yield.round(2)
```

```python
state_season_yield.stack().sort_values(ascending=False).head(10).round(2)
```

```python
Q1 = df["Profit_INR"].quantile(0.25)
Q3 = df["Profit_INR"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Profit_INR"] < lower_limit) |
    (df["Profit_INR"] > upper_limit)
]

print("Lower Limit:", round(lower_limit, 2))
print("Upper Limit:", round(upper_limit, 2))
print("Number of Outliers:", len(outliers))
```

```python
outliers[["State", "District", "Crop", "Season", "Profit_INR"]].head(10)
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["Profit_INR"])

plt.title("Distribution of Farm Profit and Outliers")
plt.xlabel("Profit (INR)")

plt.show()
```

## Seasonal Consistency Across States

The analysis shows considerable variation in yield across seasons and states.

- Punjab has the highest seasonal yield variation of 4.51 tonnes/ha.
- Maharashtra has the second-highest variation of 3.22 tonnes/ha.
- Telangana has the lowest variation of 1.30 tonnes/ha.
- Punjab shows greater dependence on seasonal conditions, while Telangana shows relatively more consistent performance across seasons.

This indicates that regional and seasonal factors can strongly influence agricultural productivity.

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    state_season_yield,
    annot=True,
    fmt=".2f"
)

plt.title("Average Yield by State and Season")
plt.xlabel("Season")
plt.ylabel("State")

plt.show()
```

```python
# Water Efficiency by Irrigation Method

irrigation_efficiency = df.groupby('Irrigation_Method')['Water_Efficiency_t_per_1000m3'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
irrigation_efficiency.plot(kind='bar')
plt.title('Water Efficiency by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Water Efficiency (t/1000 m³)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 10. Key Findings

1. **Kharif season performed best overall**, with the highest average yield of 5.63 tonnes/ha and the highest average profit of approximately ₹178,915.

2. **Zaid season showed weaker economic performance**, with an average profit of approximately -₹24,805, indicating an average loss.

3. **Sugarcane had the highest average yield across all three seasons**, while Sugarcane and Chilli were among the most profitable crops.

4. **Zaid used the highest average water** (6419.89 m³) and had the lowest water-use efficiency (4.41 tonnes per 1000 m³).

5. **Kharif had the highest rainfall, humidity, and soil moisture**, while Zaid had the highest temperature and sunlight hours.

6. **Punjab–Rabi recorded the highest state-season yield** at 8.61 tonnes/ha.

7. **Maharashtra–Zaid recorded the lowest state-season yield** at 2.28 tonnes/ha.

8. **Punjab showed the highest seasonal yield variation** (4.51 tonnes/ha), while Telangana showed the lowest variation (1.30 tonnes/ha).

9. The Kruskal-Wallis test indicated a **statistically significant difference in yield across seasons** (p < 0.05).

10. Profit was strongly associated with revenue and moderately associated with production, yield, and water efficiency. These relationships indicate association rather than causation.

11. The analysis identified **404 unusual profit observations**, including both unusually high profits and large losses.

## 11. Recommendations

Based on the analysis, the following recommendations can be made:

- Focus on high-performing crops such as Sugarcane and Chilli where conditions are suitable.
- Give special attention to Zaid season because it showed lower yield, lower water efficiency, and negative average profit.
- Improve water management during seasons with high water requirements.
- Consider regional seasonal differences when planning crop selection and resource allocation.
- Investigate the causes of unusually high profits and large losses before making major decisions.
- Use seasonal yield and profitability trends to support better agricultural planning.
- Promote efficient irrigation and resource-use practices to improve productivity and reduce unnecessary water consumption.
